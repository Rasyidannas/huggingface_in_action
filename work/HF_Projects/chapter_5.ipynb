```python
import os
from dotenv import load_dotenv
load_dotenv()

# Validating the availability of the dataset
import requests 

my_token = os.getenv("HF_TOKEN")
token = my_token

dataset_id = 'fka/prompts.chat'

headers = {"Authoriztion": f"Bearer {token}"}

API_URL = f"https://datasets-server.huggingface.co/is-valid?dataset={dataset_id}"

def query():
    response = requests.get(API_URL, headers=headers)
    return response.json()

data = query()
data

print(data)

'''
Summary here, it is just for valdate datasets and you can use curl like this:
curl -X GET "https://datasets-server.huggingface.co/is-valid?dataset=fka/prompts.chat"
'''
```

```python
# Download a dataset
from datasets import load_dataset

dataset_id = 'stanfordnlp/imdb'

dataset = load_dataset(dataset_id)
# Show the dataset structure
print(dataset)

# Get the first row of the train subset
train_first = dataset['train'][0]
print(train_first)

'''
Summary in here:
The result is represented as a DatasetDict object:
- train : The training dataset, used to train models
- test : The testing dataset, used to evaluate the model’s performance
- unsupervised : Subset that often contains unlabeled data, which can be used for unsupervised or semisupervised learning tasks

You can use curl for show the dataset structure
curl -X GET "https://datasets-server.huggingface.co/splits?dataset=stanfordnlp/imdb"
'''
```

```python
# you can donwload specific subset of dataset
dataset = load_dataset(dataset_id, split='train')

print(dataset)
```
