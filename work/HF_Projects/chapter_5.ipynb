```python
import os
from dotenv import load_dotenv
load_dotenv()

# Validating the availability of the dataset
import requests 

my_token = os.getenv("HF_TOKEN")
token = my_token

dataset_id = 'fka/prompts.chat'

headers = {"Authoriztion": f"Bearer {token}"}

API_URL = f"https://datasets-server.huggingface.co/is-valid?dataset={dataset_id}"

def query():
    response = requests.get(API_URL, headers=headers)
    return response.json()

data = query()
data

print(data)

'''
Summary here, it is just for valdate datasets and you can use curl like this:
curl -X GET "https://datasets-server.huggingface.co/is-valid?dataset=fka/prompts.chat"
'''
```
