```python
# Tokenizing datasets
from transformers import AutoTokenizer

dataset = load_dataset(dataset_id)
# Use bert-base-uncased for tokenizer
tokenizer = AutoTokenizer.from_pretrained('bert-base-uncased')
tokenized_dataset = dataset.map(
    lambda examples:
        tokenizer(examples['text'],
                  truncation = True,
                  padding = 'max_length'),
    batched = True
)

print(tokenized_dataset)

'''
Summary here:
-  token ID is a unique numerical identifier assigned to a token
(a word, subword, punctuation mark, or even whitespace, depending on the
tokenizer used) in the context of NLP.

- The second attribute, token_type_ids, is used to differentiate among multiple seg-
ments of a single input. This attribute is particularly useful for models such as BERT,
which can process inputs consisting of two separate segments (e.g., a sentence pair in
tasks like next-sentence prediction or question answering). 

- The third attribute, attention_mask, is used to inform the model which tokens
should be attended to (processed) and which should not. 
'''
```

```python
# Show input_ids
print(tokenized_dataset['train'][0]['input_ids'])
```

```python
# Convert the token IDs to tokens
tokens = tokenizer.convert_ids_to_tokens(tokenized_dataset['train'][0]['input_ids'])

print(tokens)
```

```python
# Show attention mask
print(tokenized_dataset['train'][0]['attention_mask'])
```

```python
# Visualizing datasets
from datasets import load_dataset

dataset = load_dataset('zeroshot/twitter-financial-news-topic')

# Extraxt the topic labels from the training set
train_data = dataset['train']

# Show first row of the data
print(train_data[0])
```

```python
# Show the last row in the data
print(train_data[-1])
```

```python
# Defining the list of topics 
topics = {
    "LABEL_0": "Analyst Update",
    "LABEL_1": "Fed | Central Banks",
    "LABEL_2": "Company | Product News",
    "LABEL_3": "Treasuries | Corporate Debt",
    "LABEL_4": "Dividend",
    "LABEL_5": "Earnings",
    "LABEL_6": "Energy | Oil",
    "LABEL_7": "Financials",
    "LABEL_8": "Currencies",
    "LABEL_9": "General News | Opinion",
    "LABEL_10": "Gold | Metals | Materials",
    "LABEL_11": "IPO",
    "LABEL_12": "Legal | Regulation",
    "LABEL_13": "M&A | Investments",
    "LABEL_14": "Macro",
    "LABEL_15": "Markets",
    "LABEL_16": "Politics",
    "LABEL_17": "Personnel Change",
    "LABEL_18": "Stock Commentary",
    "LABEL_19": "Stock Movement",
}

mapped_labels = [topics[f"LABEL_{label}"] 
                 for label in train_data['label']]
```

```python
# Plotting the dataset using a bar chart
%pip install matplotlib
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(10, 6))

# Create bin edges to center the labels
bins = np.arange(len(topics) + 1) - 0.5

plt.hist(mapped_labels, 
        bins = bins,
        edgecolor = 'black',
        color = 'skyblue',
        alpha = 0.7)

# Sets the ticks at the center of each bin
plt.xticks(np.arange(len(topics)),
          list(topics.values()),
          rotation = 90,
          ha = 'center')

plt.title("Topic Distribution - Twitter Financial News")
plt.xlabel("Topics")
plt.ylabel("Number of Tweets")
plt.tight_layout()
plt.show()
```

```python
# Using the CIFAR-10 dataset
import numpy as np

dataset = load_dataset('uoft-cs/cifar10')

print(dataset)
```

```python
# Access first image
dataset['train'][0]
```

```python
# Define Label for the CIFAR-10 classes
labels = {
    0: "airplane",
    1: "automobile",
    2: "bird",
    3: "cat",
    4: "deer",
    5: "dog",
    6: "frog",
    7: "horse",
    8: "ship",
    9: "truck"
}

# Visualizes a few images from the training set
def show_images(images, labels, labels_dict):
    plt.figure(figsize=(5, 5))
    for i in range(25):
        plt.subplot(5, 5, i + 1)
        plt.imshow(images[i])
        plt.title(labels_dict[labels[i]])
        plt.axis('off')
    plt.tight_layout()
    plt.show()

# Get 25 random samples
train_samples = dataset['train'].shuffle(seed=42).select(range(25))

# Extract images and labels
images = [sample['img'] for sample in train_samples]
class_labels = [sample['label'] for sample in train_samples]

# Display the images with class names
show_images(images, class_labels, labels)
```
